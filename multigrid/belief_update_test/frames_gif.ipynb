```python
from PIL import Image
import glob, os, re

def step_num(path):
    m = re.search(r'step_(\d+)', os.path.basename(path))
    return int(m.group(1)) if m else float('inf')

paths = sorted(glob.glob("actor_belief_step_*.png"), key=step_num)

frames = [Image.open(p) for p in paths]

frames[0].save(
    "actor_belief.gif",
    save_all=True,
    append_images=frames[1:],
    duration=200,   
    loop=0,
    optimize=True
)

print("GIF saved: actor_belief.gif")

```
