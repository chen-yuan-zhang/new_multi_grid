```python
import gymnasium as gym
import envs

env = gym.make('MultiGrid-AGRlocked-v0',render_mode = 'human' )
```

```python
print(type(env))               # <class 'gymnasium.wrappers.TimeLimit'>
print(type(env.unwrapped))     # <class 'multigrid.envs.agrenv.AGREnv'>
for agent in env.unwrapped.agents:
    print(type(agent))
```

```python
observations, infos = env.reset()

while not env.unwrapped.is_done():
   # this is where you would insert your policy / policies
    actions = {agent.index: agent.action_space.sample() for agent in env.unwrapped.agents}
    observations, rewards, terminations, truncations, infos = env.step(actions)
    # print("rewards: ",rewards)
    # print("terminations: ",terminations)
    # print("truncations: ", truncations)
    # print("infos: ",infos)
    
    

env.close()
```
